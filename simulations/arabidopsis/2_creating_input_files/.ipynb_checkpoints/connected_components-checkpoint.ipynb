```python
import sys
sys.path.append("/Users/jameskitchens/Documents/GitHub/terracotta")
import terracotta as tct
import importlib
importlib.reload(tct)
import pandas as pd
import networkx as nx


demes = pd.read_csv("dataset/demes.tsv", sep="\t")
connections = pd.read_csv("dataset/connections.tsv", sep="\t")
samples = pd.read_csv("dataset/samples.tsv", sep="\t")
world_map = tct.WorldMap(demes, connections, samples)
```

```python
for epoch in world_map.epochs:
    G = nx.Graph()
    for i,deme in world_map.demes.iterrows():
        if world_map.get_deme_suitability_at_time(deme["id"], epoch) > 0:
            G.add_node(deme["id"])
    for i,connection in world_map.connections.iterrows():
        s0 = world_map.get_deme_suitability_at_time(connection["deme_0"], epoch)
        s1 = world_map.get_deme_suitability_at_time(connection["deme_1"], epoch)
        if (s0 > 0) and (s1 > 0):
            G.add_edge(connection["deme_0"], connection["deme_1"])
    print(epoch, len(list(nx.connected_components(G))))
```
